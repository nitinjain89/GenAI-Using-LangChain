Structured Tool

```python
from langchain_core.tools import StructuredTool
from pydantic import BaseModel, Field
```

```python
class MultiplyInput(BaseModel):
    a: int = Field(description='First number to multiply')
    b: int = Field(description='Second number to multiply')
```

```python
def multiply_func(a:int, b:int) -> int:
    return a*b
```

```python
multiply_tool = StructuredTool.from_function(
    func=multiply_func,
    name='multiply',
    description="Multiply two numbers",
    args_schema=MultiplyInput
)
```

```python
multiply_tool.invoke({'a':5, 'b':9})
```

```python
multiply_tool.name
```

```python
multiply_tool.description
```

```python
multiply_tool.args
```

BaseTool Class

```python
from langchain.tools import BaseTool
from typing import Type
```

```python
class MultiplyInput(BaseModel):
    a: int = Field(description='First number to multiply')
    b: int = Field(description='Second number to multiply')
```

```python
class MultiplyTool(BaseTool):
    name: str = "Multiply",
    description: str = "Multiply two numbers",
    args_schema: Type[BaseModel] = MultiplyInput

    def _run(self, a: int, b: int) -> int:
        return a*b
```

```python
mult_tool = MultiplyTool()
```

```python
mult_tool.invoke({'a':10, 'b':20})
```

```python
mult_tool.name
```

```python
mult_tool.description
```

ToolKit

```python
from langchain_core.tools import tool
```

```python
@tool
def add(a: int, b: int) -> int:
    """Add two numbers"""
    return a + b

@tool
def multiply(a: int, b: int) -> int:
    """Multiply two numbers"""
    return a * b
```

```python
class MathToolkit:
    def get_tools(self):
        return [add, multiply]
```

```python
toolkit = MathToolkit()
tools = toolkit.get_tools()

for tool in tools:
    print(tool.name, "=>", tool.description)
```
