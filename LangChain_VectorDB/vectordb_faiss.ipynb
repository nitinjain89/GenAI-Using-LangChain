```python
from langchain_google_genai import GoogleGenerativeAIEmbeddings
from langchain_community.vectorstores import Chroma, FAISS
from langchain_core.documents import Document
from dotenv import load_dotenv
```

```python
load_dotenv()
```

```python
embedding_model = GoogleGenerativeAIEmbeddings(model="gemini-embedding-001", output_dimensionality=100)
```

```python
doc1 = Document(
    page_content="I started my career in 2011",
    metadata={"title": "career"}
)
doc2 = Document(
    page_content="Ericsson was my first company",
    metadata={"title": "Ericsson"}
)
doc3 = Document(
    page_content="I married to Neha in 20's",
    metadaata={'title': 'Married'}
)

doc = [doc1, doc2, doc3]
```

```python
vector_store = FAISS.from_documents(
    embedding=embedding_model,
    documents=doc
)
```

```python
vector_store.similarity_search_with_score(
    query="whom I marry?",
    k=2
)
```

```python
results = vector_store.similarity_search_with_score(
    query="whom I marry?",
    k=2
)
```

```python
for i, doc in enumerate(results):
    print(f"\nResult {i+1}:")
    print(doc)
```
