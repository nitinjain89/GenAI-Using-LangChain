```python
from langchain_community.vectorstores import Chroma
from langchain_google_genai import GoogleGenerativeAIEmbeddings
from langchain_core.documents import Document
from dotenv import load_dotenv
```

```python
load_dotenv()
```

```python
documents = [
    Document(page_content="Python is a popular programming language."),
    Document(page_content="Python is widely used for machine learning."),
    Document(page_content="Machine learning helps computers learn from data."),
    Document(page_content="Deep learning uses neural networks."),
    Document(page_content="Neural networks are inspired by the human brain.")
]
```

```python
embedding_model = GoogleGenerativeAIEmbeddings(model="gemini-embedding-001", output_dimensionality=100)
```

```python
vector_store = Chroma.from_documents(
    documents=documents,
    embedding=embedding_model,
    collection_name="my_collection_nj"
)
```

```python
retriever = vector_store.as_retriever(search_kwargs={"k":2})
```

```python
query = "what are Neural Networks?"
results = retriever.invoke(query)
```

```python
results
```

```python
for i, doc in enumerate(results):
    print(f"\n-- Result {i+1} --")
    print(doc.page_content)
```
