```python
from langchain_core.documents import Document
```

```python
sentences = [
    "Python is widely used for data science and machine learning.",
    "Python provides libraries like NumPy and Pandas for data analysis.",
    "Python is a popular programming language for AI development.",
    "Machine learning enables computers to learn patterns from data.",
    "Deep learning uses neural networks to solve complex problems.",
    "LangChain helps developers build applications using Large Language Models.",
    "RAG retrieves relevant documents before generating answers using LLMs.",
    "Embeddings convert text into numerical vector representations.",
    "FAISS is a library used for efficient similarity search over vectors.",
    "MMR retrieves relevant documents while reducing redundant results."
]
```

```python
documents = []

for i, sentence in enumerate(sentences):
    doc = Document(
            page_content=sentence,
            metadata={"id":i+1}
        )
    documents.append(doc)
             
```

```python
from langchain_community.vectorstores import FAISS
from langchain_google_genai import GoogleGenerativeAIEmbeddings
from dotenv import load_dotenv
```

```python
load_dotenv()

embeddings = GoogleGenerativeAIEmbeddings(model="gemini-embedding-001", output_dimensionality=100)
```

```python
vectorstore = FAISS.from_documents(
    documents=documents,
    embedding=embeddings
)
```

```python
retriever = vectorstore.as_retriever(
    search_type="mmr",
    search_kwargs={"k": 3, "lambda_mult": 0.5}
)
```

```python
query = "what is python?"
retriever.invoke(query)
```
