```python
from langchain_groq import ChatGroq
from langchain_core.tools import tool
from langchain.agents import create_agent
from langchain_tavily import TavilySearch
import requests
```

```python
llm = ChatGroq(model='qwen/qwen3.8-27b')
```

```python
search_tool = TavilySearch()

@tool
def get_weather_data(city: str) -> str:
    """
    This function fetches the current weather data for a given city
    """
    url = f"https://api.apilayer.net/weatherstack/current?access_key=3e306cb4e5628efc55c4a95eaa35a901&query={city}"
    response = requests.get(url)

    return response.json()
```

```python
from langsmith import Client

client = Client()

prompt = client.pull_prompt("hwchase17/react", dangerously_pull_public_prompt=True)
```

```python
system_prompt = """
You are a helpful AI assistant.

You have access to the following tools:
1. Search tool: Search the internet for information.
2. Weather tool: Get current weather data for a given city.

Use the appropriate tool whenever required.
Provide accurate and concise answers.
"""
```

```python
agent = create_agent(
    model=llm,
    tools=[search_tool, get_weather_data],
    system_prompt=system_prompt
)
```

```python
response = agent.invoke({'messages': "What is the capital of Rajasthan and it's current weather condition?"})
print(response['messages'][-1].content)
```
